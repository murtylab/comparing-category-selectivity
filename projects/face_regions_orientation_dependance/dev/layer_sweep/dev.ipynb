```python
%load_ext autoreload
%autoreload 2

import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge

from compcats.datasets.fmri import nsd
from compcats.models import resnet50_openaiclip
from compcats.utils import model_utils
from compcats.encoder import encode

NSD_DIR = '/Users/adipani3/datasets/nsd'
MODELS_DIR = '../../../preprint/models'
SEED = 42
```

```python
subject = 1
froi = 'rFFA'
tthreshold_dict = {
    'rFFA': 6,
    'rOFA': 3
}
train_subset = 485
test_subset = 515

model = resnet50_openaiclip.get_model()
layer = 'layer4.1.conv2' # Example layer name
X_trainval = model_utils.get_dataset_activations(
    MODELS_DIR, NSD_DIR, model, layer, 'nsd', subset=train_subset
)
X_trainval = X_trainval.flatten(start_dim=1).cpu().numpy()
Y_trainval, _ = nsd.get_betas(
    NSD_DIR, 
    subject,
    froi,
    tthreshold_dict[froi],
    subset=train_subset,
    voxels=True
)

mapper = Ridge()
results = encode.get_optimal_model(
    mapper, 
    X_trainval,
    Y_trainval,
    seed=SEED
)
```
