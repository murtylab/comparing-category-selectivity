```python
%load_ext autoreload
%autoreload 2

import os
import numpy as np
import sys
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "../preprint")))

from preprint import paths

from compcats.datasets.fmri import murty185
from compcats.models import resnet50_imagenet1kv2
from compcats.utils import model_utils

MODELS_DIR = '../preprint/models'
```

```python
froi = 'FFA'
subject = 1
model = resnet50_imagenet1kv2.get_model()
layer = 'layer4.1.conv2'
```

```python
betas, _ = murty185.get_betas(
    paths.MURTY185_DIR,
    subject=subject,
    froi=froi,
    voxels=True
)
Y_train = betas[:100, :]
Y_test = betas[100:, :]
```

```python
model_acts = model_utils.get_dataset_activations(
    MODELS_DIR, paths.MURTY185_DIR, model, layer, 'murty185', 
    froi=froi, device='mps'
)
X_train = model_acts[:100, :].cpu().numpy()
X_test = model_acts[100:, :].cpu().numpy()
```

```python
assert X_train.shape[0] == Y_train.shape[0], "Number of training samples must match"
assert X_test.shape[0] == Y_test.shape[0], "Number of test samples must match"
```

```python
save_dir = './data/intro_to_ems'
os.makedirs(save_dir, exist_ok=True)
np.save(os.path.join(save_dir, 'X_train.npy'), X_train)
np.save(os.path.join(save_dir, 'Y_train.npy'), Y_train)
np.save(os.path.join(save_dir, 'X_test.npy'), X_test)
np.save(os.path.join(save_dir, 'Y_test.npy'), Y_test)
```
