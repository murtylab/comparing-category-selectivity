```python
import numpy as np
import pandas as pd
import scipy
```

```python
RESULTS_DIR = 'results'
MODELS_DIR = 'models'

LOCALIZATION_VALIDATION_CSV_DIR = f'{RESULTS_DIR}/localization_validation_konkle.csv'
SUBJECTSUBJECT_PEARSONR_CSV_PATH = f'{RESULTS_DIR}/subjectsubject_pearsonr_nsd515.csv'
MODELSUBJECT_PEARSONR_CSV_PATH = f'{RESULTS_DIR}/modelsubject_pearsonr_nsd515.csv'
```

## Localization Validation

```python
localization_validation_csv = pd.read_csv(LOCALIZATION_VALIDATION_CSV_DIR)
localization_validation_tvalues = localization_validation_csv['tvalue']
localization_validation_pvalues = localization_validation_csv['pvalue']

print(f'Median t-value: {np.median(localization_validation_tvalues)}')
print(f'Mean t-value: {np.mean(localization_validation_tvalues)}')
print(f'Maximum t-value: {np.max(localization_validation_tvalues)}')
print(f'Minimum t-value: {np.min(localization_validation_tvalues)}')

print(f'Maximum p-value: {np.max(localization_validation_pvalues)}')
```

# Subject-Subject Univariate Correlation

```python
subjectsubject_univariate_csv = pd.read_csv(SUBJECTSUBJECT_PEARSONR_CSV_PATH)
subjectsubject_univariate_pearsonr = subjectsubject_univariate_csv['pearsonr']
subjectsubject_univariate_pvalues = subjectsubject_univariate_csv['pvalue']
print(f'Median across all subjects: {subjectsubject_univariate_pearsonr.median()}')
print(f'Mean across all subjects: {subjectsubject_univariate_pearsonr.mean()}')
print(f'Minimum across all subjects: {subjectsubject_univariate_pearsonr.min()}')
print(f'Maximum across all subjects: {subjectsubject_univariate_pearsonr.max()}')

print(f'Maximum p-value across all subjects: {subjectsubject_univariate_pvalues.max()}')
```

# Model-Subject Univariate Correlation

```python
subjectsubject_univariate_csv = pd.read_csv(SUBJECTSUBJECT_PEARSONR_CSV_PATH)
modelsubject_univariate_csv = pd.read_csv(MODELSUBJECT_PEARSONR_CSV_PATH)

frois = subjectsubject_univariate_csv['fmri_froi'].unique()
subjects = subjectsubject_univariate_csv['subject1'].unique()
models = modelsubject_univariate_csv['model_name'].unique()

for froi in ['FFA', 'PPA', 'EBA']:
    hh_means = []
    for subject in subjects:
        subjectsubject_froi = subjectsubject_univariate_csv[
            (subjectsubject_univariate_csv['fmri_froi'] == froi) &
            (subjectsubject_univariate_csv['subject1'] == subject)
        ]
        hh_means.append(subjectsubject_froi['pearsonr'].mean())
    hh_means = np.array(hh_means)

    mh_means = []
    for model in models:
        modelsubject_froi = modelsubject_univariate_csv[
            (modelsubject_univariate_csv['fmri_froi'] == froi) &
            (modelsubject_univariate_csv['model_froi'] == froi) &
            (modelsubject_univariate_csv['model_name'] == model)
        ]
        mh_means.append(modelsubject_froi['pearsonr'].mean())
    mh_means = np.array(mh_means)

    tvalue, pvalue = scipy.stats.ttest_ind(
        hh_means, mh_means, equal_var=False, nan_policy='raise'
    )
    print(f'T-test for {froi}: t-value = {tvalue}, p-value = {pvalue}')
```
