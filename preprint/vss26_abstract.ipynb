```python
from IPython.display import display
import pandas as pd

import main_analysis as ma

RESULTS_DIR = 'results'
SUBJECTSUBJECT_CORRELATIONS_CSV_PATH = f'{RESULTS_DIR}/subjectsubject_correlations_nsd515.csv'
MODELSUBJECT_CORRELATIONS_CSV_PATH = f'{RESULTS_DIR}/modelsubject_correlations_nsd515.csv'
```

```python
core_frois = ['FFA', 'PPA', 'EBA']
all_frois = ['FFA', 'OFA', 'PPA', 'OPA', 'EBA', 'RSC', 'FBA', 'VWFA', 'OVWFA']
frois = core_frois
tthreshold = 7

to_model_names = ma.get_model_names('task_optimized')
model_localizer = 'grillspector'
best_model = 'vitb16siglip2res224'
metrics = ['pearsonr', 'rdmr'] # 'normed_pearsonr' 'pearsonr'
```

## Subject-Subject correlation

```python
for metric in metrics:
    print(f'\n=== SUBJECT-SUBJECT correlations: {metric} ===')
    sbjsbj_csv = pd.read_csv(SUBJECTSUBJECT_CORRELATIONS_CSV_PATH)
    sbjsbj_csv = sbjsbj_csv[
        (sbjsbj_csv['fmri_froi'].isin(frois)) &
        (sbjsbj_csv['fmri_tthreshold'] == tthreshold)
    ]
    sbjsbj_csvgrouped = sbjsbj_csv.groupby('fmri_froi')[metric].median().reset_index()
    display(sbjsbj_csvgrouped)
    print(sbjsbj_csvgrouped[metric].median())
```

# Model-Subject correlation

```python
for metric in metrics:
    print(f'\n=== MODEL-SUBJECT correlations: {metric} ===')
    mdlsbj_csv = pd.read_csv(MODELSUBJECT_CORRELATIONS_CSV_PATH)
    mdlsbj_csv = mdlsbj_csv[
        (mdlsbj_csv['model_name'].isin(to_model_names)) &
        (mdlsbj_csv['fmri_froi'].isin(frois)) &
        (mdlsbj_csv['fmri_tthreshold'] == tthreshold) &
        (mdlsbj_csv['model_froi'].isin(frois)) &
        (mdlsbj_csv['model_tthreshold'] == tthreshold) &
        (mdlsbj_csv['model_localizer'] == model_localizer) &
        (mdlsbj_csv['model_name'].isin(to_model_names))
    ]
    # print(mdlsbj_csv[metric].median())
    mdlsbj_csvgrouped = mdlsbj_csv.groupby(['model_name', 'fmri_froi'])[metric].median().reset_index()
    mdlsbj_csvgrouped2 = mdlsbj_csvgrouped.groupby('fmri_froi')[metric].median().reset_index()
    print(mdlsbj_csvgrouped2[metric].median())


# best_mdlsbj_csv = mdlsbj_csv[
#     (mdlsbj_csv['model_name'].isin(to_model_names)) &
#     (mdlsbj_csv['fmri_froi'].isin(frois)) &
#     (mdlsbj_csv['fmri_tthreshold'] == tthreshold) &
#     (mdlsbj_csv['model_froi'].isin(frois)) &
#     (mdlsbj_csv['model_tthreshold'] == tthreshold) &
#     (mdlsbj_csv['model_localizer'] == model_localizer) &
#     (mdlsbj_csv['model_name'] == best_model)
# ]
# best_mdlsbj_csvgrouped = best_mdlsbj_csv.groupby(['model_name', 'fmri_froi'])[metric].median().reset_index()
# best_mdlsbj_csvgrouped2 = best_mdlsbj_csvgrouped.groupby('fmri_froi')[metric].median().reset_index()
# print(best_mdlsbj_csvgrouped2[metric].median())
```
